```python
# Define a helper function to normalize URLs so trailing slashes or uppercase letters do not cause duplicates
def normalize_url(url):
    if not url or not isinstance(url, str):
        return ""
    # Strip surrounding whitespace, convert to lowercase, and remove any trailing forward slash
    cleaned_url = url.strip().lower().rstrip("/")
    # Return the standardized URL string
    return cleaned_url

# Define a helper function to clean raw extracted webpage text
def clean_text(text):
    if not text or not isinstance(text, str):
        # Return empty string for invalid text inputs
        return ""
    # Replace Windows-style carriage returns with standard newline characters
    text = text.replace("\r\n", "\n")
    # addzero factual values
    lines = [line.strip() for line in text.split("\n") if line.strip() and line.strip().lower() != "linkedin"]
    cleaned = "\n".join(lines)
    cleaned = re.sub(r"\n{3,}", "\n\n", cleaned)
    # Collapse multiple horizontal spaces or tabs into a single space
    cleaned = re.sub(r"[ \t]+", " ", cleaned)
    # Return the final stripped text string
    return cleaned.strip()

#deduplicated webpage records
cleaned_records = []
seen_urls = set()
seen_text_hashes = set()
skipped_invalid = 0
skipped_duplicate_urls = 0
skipped_duplicate_content = 0

# Loop through each record in the raw dataset along with its index
for idx, item in enumerate(raw_banoqabil_data):
    if not isinstance(item, dict):
        skipped_invalid += 1
        continue

    # Extract and normalize the page URL
    norm_url = normalize_url(item.get("url", ""))
    title = str(item.get("title", "Untitled Page")).strip()
    cleaned_body = clean_text(item.get("text", ""))

    # Skip records that lack a valid URL or have fewer than 30 characters of meaningful text
    if not norm_url or len(cleaned_body) < 30:
        skipped_invalid += 1
        continue
    if norm_url in seen_urls:
        skipped_duplicate_urls += 1
        continue

    # Compute a SHA-256 hash of the cleaned text to detect identical content across different URLs
    content_hash = hashlib.sha256(cleaned_body.encode("utf-8")).hexdigest()

    # Check whether identical text content has already been saved from another URL
    if content_hash in seen_text_hashes:
        skipped_duplicate_content += 1
        continue
    seen_urls.add(norm_url)
    seen_text_hashes.add(content_hash)

    # Append the clean, standardized dictionary to our final list
    cleaned_records.append({
        "doc_id": f"bq_doc_{len(cleaned_records) + 1:03d}",
        "url": norm_url,
        "title": title,
        "text": cleaned_body,
        "char_count": len(cleaned_body)
    })

# Define the output file path inside Colab's data folder
output_path = os.path.join("data", "cleaned_banoqabil.json")

# Open the output JSON file in write mode with UTF-8 encoding
with open(output_path, "w", encoding="utf-8") as f:
    # unicode char
    json.dump(cleaned_records, f, ensure_ascii=False, indent=2)

# REPORT
print("--- Stage 1 Preprocessing Summary ---")
print(f"Total raw records:          {len(raw_banoqabil_data)}")
print(f"Clean unique pages saved:   {len(cleaned_records)}")
print(f"Duplicate URLs removed:     {skipped_duplicate_urls}")
print(f"Duplicate content removed:  {skipped_duplicate_content}")
print(f"Invalid/empty skipped:      {skipped_invalid}")
print(f"Saved to:                   {output_path}")
```

Context-Aware Text Chunking (RecursiveCharacterTextSplitter): Large pages like the 31,690-character faculty directory cannot be embedded as a single block because embedding models lose fine-grained detail over long texts and LLMs retrieve noise. Splitting pages into 800-character chunks with 150-character overlap (and prepending the page title to every chunk) keeps individual course descriptions, instructor bios, and admission steps self-contained.

Hosted Embedding Model (gemini-embedding-001 via google-genai): Instead of downloading a heavy local transformer model onto Colab's CPU, we send batches of text chunks to Google's hosted embedding API using task_type="RETRIEVAL_DOCUMENT". We configure output_dimensionality=768 so the vectors are fast, compact, and match our Stage 3 Pinecone index configuration cleanly.

```python

cleaned_file_path = os.path.join("data", "cleaned_banoqabil.json")

# Open and load the 42 cleaned BanoQabil records from disk
with open(cleaned_file_path, "r", encoding="utf-8") as f:
    cleaned_records = json.load(f)
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=800,
    # Set 150-character overlap so sentences split across boundaries keep their context
    chunk_overlap=150,
    separators=["\n\n", "\n", ". ", " ", ""]
)
chunked_records = []
for doc in cleaned_records:
    raw_chunks = text_splitter.split_text(doc["text"])
    for chunk_idx, chunk_text in enumerate(raw_chunks):
        contextualized_text = f"Page Title: {doc['title']}\n{chunk_text}"
        chunk_id = f"{doc['doc_id']}_chunk_{chunk_idx:03d}"
        chunked_records.append({
            "id": chunk_id,
            "doc_id": doc["doc_id"],
            "url": doc["url"],
            "title": doc["title"],
            "chunk_index": chunk_idx,
            "text": contextualized_text
        })
chunks_output_path = os.path.join("data", "chunked_banoqabil.json")

# Save the chunked dataset to disk for inspection and backup
with open(chunks_output_path, "w", encoding="utf-8") as f:
    json.dump(chunked_records, f, ensure_ascii=False, indent=2)
print(f"Split {len(cleaned_records)} pages into {len(chunked_records)} chunks.")
print(f"\nSample Chunk [{chunked_records[0]['id']}]:\n{chunked_records[0]['text'][:300]}...")
```

```python

EMBEDDING_MODEL = "gemini-embedding-001"
EMBEDDING_DIM = 768
BATCH_SIZE = 40
embedded_records = []
start_idx = 0
print(f"Generating {EMBEDDING_DIM}-dim embeddings for {len(chunked_records)} chunks (paced for Free Tier limits)...")

# Loop until every chunk in chunked_records has been embedded
while start_idx < len(chunked_records):
    batch = chunked_records[start_idx : start_idx + BATCH_SIZE]
    batch_texts = [item["text"] for item in batch]

    # Wrap the API call in a try/except block to automatically recover from 429 rate limits
    try:
        # Call the hosted Gemini embedding API for the current batch
        response = genai_client.models.embed_content(
            # Specify the embedding model
            model=EMBEDDING_MODEL,
            # Pass the list of text chunks to embed
            contents=batch_texts,
            # Configure task_type for document indexing and set output dimensionality to 768
            config=types.EmbedContentConfig(
                # Optimize embeddings specifically for stored knowledge-base documents
                task_type="RETRIEVAL_DOCUMENT",
                # Fix vector length to 768 dimensions
                output_dimensionality=EMBEDDING_DIM
            )
        )

        # Pair each chunk dictionary in the batch with its returned embedding vector
        for item, embedding_obj in zip(batch, response.embeddings):
            # Append a new dictionary containing the chunk metadata and float vector values
            embedded_records.append({
                # Preserve the unique chunk ID
                "id": item["id"],
                # Store the embedding vector as a standard Python list of floats
                "values": embedding_obj.values,
                # Store all metadata fields required for Pinecone filtering and RAG citations
                "metadata": {
                    # Store the parent document ID
                    "doc_id": item["doc_id"],
                    # Store the source webpage URL
                    "url": item["url"],
                    # Store the source webpage title
                    "title": item["title"],
                    # Store the chunk sequence index
                    "chunk_index": item["chunk_index"],
                    "text": item["text"]
                }
            })

        # Advance the starting index pointer to the next batch
        start_idx += BATCH_SIZE
        # Print progress after each completed batch
        print(f"Embedded {len(embedded_records)} / {len(chunked_records)} chunks...")

        # Pause 30 seconds between batches if more chunks remain so we never exceed 100 items/minute
        if start_idx < len(chunked_records):
            # Wait 30 seconds before sending the next batch of 40 chunks
            time.sleep(30)

    # Catch API errors such as 429 RESOURCE_EXHAUSTED rate-limit responses
    except Exception as e:
        # Check if the error is a temporary rate-limit (429) error
        if "429" in str(e) or "RESOURCE_EXHAUSTED" in str(e):
            # Inform the user that the script is pausing automatically before retrying the batch
            print("Rate limit reached—waiting 45 seconds before retrying current batch...")
            # Sleep for 45 seconds to let the per-minute Free Tier quota reset
            time.sleep(45)
        # Re-raise any non-rate-limit error immediately so we can inspect it
        else:
            # Raise unexpected errors
            raise e

# Define the output file path for saving the embedded vectors locally in Colab
embeddings_output_path = os.path.join("data", "embedded_banoqabil.json")

# Save the embedded records to disk so you do not need to re-run API calls if Colab restarts
with open(embeddings_output_path, "w", encoding="utf-8") as f:
    # Write the vector records to disk in JSON format
    json.dump(embedded_records, f)
print("\n--- Stage 2 Embedding Summary ---")
print(f"Total embedded chunks: {len(embedded_records)}")
print(f"Vector dimensions:     {len(embedded_records[0]['values'])}")
print(f"Saved to:              {embeddings_output_path}")
```

```python
# Define the Pinecone index name (must be lowercase alphanumeric and hyphens only)
INDEX_NAME = "banoqabil-rag-index"
# Define the namespace inside the index to organize BanoQabil website vectors cleanly
NAMESPACE = "banoqabil-web"
# Define vector dimensions matching Stage 2 embeddings
VECTOR_DIM = 768

# Retrieve a list of existing index names in your Pinecone account
existing_indexes = [idx_info["name"] for idx_info in pc.list_indexes()]

# Create the Pinecone index only if it does not already exist
if INDEX_NAME not in existing_indexes:
    # Print status message before creating the serverless index
    print(f"Creating Pinecone index '{INDEX_NAME}' ({VECTOR_DIM} dims, cosine metric)...")
    # Create a serverless Pinecone index compatible with the free tier
    pc.create_index(
        # Set the index name
        name=INDEX_NAME,
        # Set the exact vector dimensionality (768)
        dimension=VECTOR_DIM,
        # Use cosine similarity for semantic text retrieval
        metric="cosine",
        # Configure AWS us-east-1 serverless hosting
        spec=ServerlessSpec(
            # Specify AWS as the cloud provider
            cloud="aws",
            # Specify us-east-1 region (supported on free Starter plan)
            region="us-east-1"
        )
    )
    # Wait until the index finishes initializing and reports ready status
    while not pc.describe_index(INDEX_NAME).status["ready"]:
        # Pause 2 seconds between readiness checks
        time.sleep(2)
    # Confirm that the new index is ready for vector uploads
    print(f"Index '{INDEX_NAME}' created and ready!")
# Handle the case where the index already exists from an earlier run
else:
    # Inform the user that the existing index will be reused
    print(f"Index '{INDEX_NAME}' already exists—connecting directly.")

# Connect to the target Pinecone index handle
index = pc.Index(INDEX_NAME)
print("Initial Index Stats:", index.describe_index_stats())
```

Query Embedding (task_type="RETRIEVAL_QUERY"): When a user asks a question, we embed their question using gemini-embedding-001 with task_type="RETRIEVAL_QUERY" and output_dimensionality=768. Using RETRIEVAL_QUERY (paired with RETRIEVAL_DOCUMENT from Stage 2) optimizes asymmetric semantic matching between short user questions and longer webpage chunks.

Top-K Semantic Retrieval + Similarity Threshold: We query Pinecone for the top_k=5 most relevant chunks and apply a minimum similarity threshold (MIN_SCORE = 0.45) so unrelated queries (like weather or sports) are filtered out before reaching the LLM.

Strict Grounded Generation (gemini-2.5-flash): We use temperature=0.0 and a strict system prompt that forces the assistant to answer exclusively from the retrieved BanoQabil context, cite source URLs, and explicitly state "I could not find this information in the available BanoQabil website data." whenever the context lacks the answer.

```python

EMBEDDING_MODEL = "gemini-embedding-001"
EMBEDDING_DIM = 768
INDEX_NAME = "banoqabil-rag-index"
NAMESPACE = "banoqabil-web"
TOP_K = 5
MIN_SIMILARITY_SCORE = 0.45

# Define the single canonical refusal string used across the pipeline
REFUSAL_MESSAGE = "I could not find this information in the available BanoQabil website data."
index = pc.Index(INDEX_NAME)

# Initialize a timeout-protected GenAI client (25 seconds max per request)
fast_client = genai.Client(
    # Reuse the verified Gemini API key
    api_key=gemini_api_key,
    # Set HTTP timeout to 25000 milliseconds (25 seconds)
    http_options=types.HttpOptions(timeout=25000)
)

# Prioritize high-quota Free Tier models so daily 20-RPD limits on 3.8-flash never block you
FAST_MODELS = ["gemini-3.5-flash-lite", "gemini-3.6-flash", "gemma-3-27b-it"]

# Strict anti-hallucination system instructions
SYSTEM_INSTRUCTIONS = f"""You are the official BanoQabil AI Assistant.
Your ONLY knowledge source is the 'RETRIEVED BANOQABIL CONTEXT' provided in each user prompt.

Strict Anti-Hallucination Rules:
1. Answer ONLY using facts explicitly stated in 'RETRIEVED BANOQABIL CONTEXT'.
2. Never use outside knowledge, assumptions, inferences, or guesses.
3. If the answer is not explicitly stated in 'RETRIEVED BANOQABIL CONTEXT', respond ONLY with:
   "{REFUSAL_MESSAGE}"
4. Do NOT write a 'Sources:' section or invent URLs yourself; verified source links are attached automatically by the system.
5. Keep responses clear, concise, and well-structured."""

# Retrieve relevant chunks from Pinecone above the minimum similarity threshold
def retrieve_banoqabil_context(user_query, top_k=TOP_K, min_score=MIN_SIMILARITY_SCORE):
    # Embed user query with RETRIEVAL_QUERY task type
    query_embedding_response = genai_client.models.embed_content(
        model=EMBEDDING_MODEL,
        contents=user_query,
        config=types.EmbedContentConfig(
            task_type="RETRIEVAL_QUERY",
            output_dimensionality=EMBEDDING_DIM
        )
    )
    # Extract float vector
    query_vector = query_embedding_response.embeddings[0].values

    # Query Pinecone namespace for top matching chunks
    search_results = index.query(
        namespace=NAMESPACE,
        vector=query_vector,
        top_k=top_k,
        include_metadata=True
    )

    # Keep only chunks that pass the similarity threshold and contain non-empty text
    relevant_chunks = []
    for match in search_results.get("matches", []):
        chunk_text = match["metadata"].get("text", "").strip()
        if match["score"] >= min_score and chunk_text:
            relevant_chunks.append({
                "id": match["id"],
                "score": round(match["score"], 4),
                "title": match["metadata"].get("title", "Unknown Title"),
                "url": match["metadata"].get("url", "").strip(),
                "text": chunk_text
            })
    return relevant_chunks

# Call Gemini with automatic model fallback and timeout protection
def generate_with_fallback(prompt_text):
    last_error = None
    for model_name in FAST_MODELS:
        try:
            response = fast_client.models.generate_content(
                model=model_name,
                contents=prompt_text,
                config=types.GenerateContentConfig(
                    system_instruction=SYSTEM_INSTRUCTIONS,
                    temperature=0.0
                )
            )
            return response.text.strip(), model_name
        except Exception as e:
            last_error = e
            print(f"    [Notice] {model_name} busy/timed out, switching model...")
            time.sleep(2)
    raise last_error

# Main grounded RAG chatbot function with Python-verified citations
def ask_banoqabil_bot(user_query, show_sources=True):
    # Retrieve matching chunks from Pinecone
    retrieved_chunks = retrieve_banoqabil_context(user_query)

    # Immediate refusal if no chunk passes the similarity threshold
    if not retrieved_chunks:
        if show_sources:
            print("=" * 70)
            print(f"Question: {user_query}")
            print("-" * 70)
            print(REFUSAL_MESSAGE)
            print("=" * 70)
        return {
            "answer": REFUSAL_MESSAGE,
            "chunks": [],
            "model_used": "none (threshold gate)"
        }

    # Build numbered context blocks from retrieved chunks
    context_blocks = [
        f"[Context {idx} | Title: {chunk['title']}]\n{chunk['text']}"
        for idx, chunk in enumerate(retrieved_chunks, start=1)
    ]
    formatted_context = "\n\n---\n\n".join(context_blocks)
    grounded_prompt = (
        f"RETRIEVED BANOQABIL CONTEXT:\n{formatted_context}\n\n"
        f"USER QUESTION:\n{user_query}"
    )

    # Generate factual answer using fallback chain
    raw_answer, used_model = generate_with_fallback(grounded_prompt)

    # Anti-hallucination guard: if the model signals missing info, return exact refusal without URLs
    refusal_markers = [
        REFUSAL_MESSAGE.lower(),
        "not found in the available banoqabil",
        "does not contain enough information",
        "not mentioned in the provided context"
    ]
    if any(marker in raw_answer.lower() for marker in refusal_markers):
        final_answer = REFUSAL_MESSAGE
    else:
        # Attach only deduplicated, non-empty URLs present in retrieved chunk metadata
        seen_sources = []
        for c in retrieved_chunks:
            if c["url"]:
                source_pair = (c["title"], c["url"])
                if source_pair not in seen_sources:
                    seen_sources.append(source_pair)
        if seen_sources:
            sources_md = "\n".join([f"* {title} ({url})" for title, url in seen_sources])
            final_answer = f"{raw_answer}\n\n**Sources:**\n{sources_md}"
        else:
            final_answer = raw_answer

    if show_sources:
        print("=" * 70)
        print(f"Question: {user_query}  [Model: {used_model}]")
        print("-" * 70)
        print(final_answer)
        print("=" * 70)

    return {
        "answer": final_answer,
        "chunks": retrieved_chunks,
        "model_used": used_model
    }

print(f"Stage 4 Chatbot Pipeline ready with models: {FAST_MODELS}")
```

Automated Evaluation Benchmark Suite: Instead of manually guessing whether your chatbot is accurate, we create a structured test suite covering four real-world query types:

Direct Fact Lookup (e.g., security deposit amount & attendance rule)

Multi-Chunk Synthesis (e.g., AI & programming courses across chunks)

Out-of-Scope / Trap Question (e.g., Aga Khan University MBBS fee — testing whether the bot refuses to invent answers)

Adversarial / Partial-Match Trap (asking about something BanoQabil doesn't offer—like a PhD degree or hostel accommodation—to verify the LLM doesn't hallucinate even when BanoQabil pages are retrieved!)

LLM-as-a-Judge Faithfulness & Relevance Grader: We use a second, independent LLM evaluation call (temperature=0.0) that inspects each (Question, Retrieved Context, Bot Answer) triple and grades:

Retrieval Relevance (PASS/FAIL): Did Pinecone retrieve chunks relevant to the question?

Groundedness / Zero Hallucination (PASS/FAIL): Is every single claim in the answer 100% supported by the retrieved context (or properly refused when missing)?

```python
# Define 5 evaluation test cases covering facts, courses, leadership, out-of-scope, and hallucination traps
evaluation_test_cases = [
    {
        "category": "Direct Fact",
        "question": "What is the refundable security deposit amount and minimum attendance required for a refund?",
        "must_contain_all": ["3,000", "60%"],
        "expect_refusal": False
    },
    {
        "category": "Course Catalog",
        "question": "Does BanoQabil offer a Generative AI course, and what topics does it cover?",
        "must_contain_all": ["Generative AI", "RAG"],
        "expect_refusal": False
    },
    {
        "category": "Organization & Leadership",
        "question": "Who is the Patron-in-Chief and Chairman of BanoQabil Pakistan, and which foundation launched it?",
        "must_contain_all": ["Alkhidmat", "Naeem", "Naveed"],
        "expect_refusal": False
    },
    {
        "category": "Out-of-Scope Trap",
        "question": "What is the tuition fee for an MBBS degree at Aga Khan University?",
        "must_contain_all": [],
        "expect_refusal": True
    },
    {
        "category": "Hallucination Trap",
        "question": "Does BanoQabil provide free on-campus hostel accommodation and monthly cash stipends to students?",
        "must_contain_all": [],
        "expect_refusal": True
    }
]

# Initialize evaluation results and exact refusal check string
eval_results = []
REFUSAL_TEXT = "i could not find this information in the available banoqabil website data."

print("Running Stage 5 Evaluation Suite (5 tests)...\n")

for idx, test in enumerate(evaluation_test_cases, start=1):
    # Query the Stage 4 chatbot
    rag_output = ask_banoqabil_bot(test["question"], show_sources=False)
    ans_lower = rag_output["answer"].lower()
    top_score = rag_output["chunks"][0]["score"] if rag_output["chunks"] else 0.0

    # Verify refusal vs. grounded factual response and source citation presence
    if test["expect_refusal"]:
        passed = REFUSAL_TEXT in ans_lower
        note = (
            "Correctly refused unsupported/out-of-scope query without hallucinating."
            if passed
            else "FAIL: Model attempted to answer an unsupported question."
        )
    else:
        missing = [kw for kw in test["must_contain_all"] if kw.lower() not in ans_lower]
        has_sources = "**sources:**" in ans_lower
        passed = (len(missing) == 0) and has_sources
        note = (
            "All required facts and verified source URLs present."
            if passed
            else f"FAIL: Missing facts={missing}, has_sources={has_sources}"
        )

    status = "PASS" if passed else "FAIL"
    print(f"[Test {idx}/5 | {test['category']}] -> {status} (Score: {top_score} | Model: {rag_output['model_used']})")
    print(f"  Note: {note}")
    print(f"  Answer Preview: {rag_output['answer'][:150].replace(chr(10), ' ')}...\n")

    eval_results.append({
        "test_id": idx,
        "category": test["category"],
        "question": test["question"],
        "top_score": top_score,
        "model_used": rag_output["model_used"],
        "status": status,
        "note": note,
        "answer": rag_output["answer"]
    })
    time.sleep(3)

# Save full evaluation report to disk
eval_output_path = os.path.join("data", "stage5_evaluation_report.json")
with open(eval_output_path, "w", encoding="utf-8") as f:
    json.dump(eval_results, f, ensure_ascii=False, indent=2)

passed_count = sum(1 for r in eval_results if r["status"] == "PASS")
print("=" * 70)
print(f"STAGE 5 EVALUATION SCORECARD: {passed_count} / {len(eval_results)} Tests Passed")
print(f"Full report saved to: {eval_output_path}")
print("=" * 70)
```
